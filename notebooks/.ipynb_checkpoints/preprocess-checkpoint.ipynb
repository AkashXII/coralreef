```python
import cv2, os, torch, numpy as np, matplotlib.pyplot as plt
import torchvision.models as models
import torchvision.transforms as T
import torch.nn as nn
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
from xgboost import XGBClassifier

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

train_transform = T.Compose([
    T.ToPILImage(),
    T.Resize((224, 224)),
    T.RandomHorizontalFlip(),
    T.ToTensor(),
    T.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

val_transform = T.Compose([
    T.ToPILImage(),
    T.Resize((224, 224)),
    T.ToTensor(),
    T.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

def load_images(base_dir):
    images, labels = [], []
    for folder, label in [("healthy", 0), ("bleached", 1)]:
        path = os.path.join(base_dir, folder)
        for fname in os.listdir(path):
            if fname.lower().endswith((".jpg", ".jpeg", ".png")):
                img = cv2.imread(os.path.join(path, fname))
                if img is None:
                    continue
                images.append(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
                labels.append(label)
    return images, labels

def build_model():
    model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
    for name, param in model.named_parameters():
        if "layer4" not in name and "fc" not in name:
            param.requires_grad = False
    model.fc = nn.Linear(model.fc.in_features, 1)
    return model.to(DEVICE)

def train_model(model, train_loader, val_loader, epochs=30):
    loss_fn = nn.BCEWithLogitsLoss()
    optimizer = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=3e-4)
    best_acc = 0.0

    for epoch in range(epochs):
        model.train()
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
            optimizer.zero_grad()
            loss = loss_fn(model(imgs).squeeze(1), labels)
            loss.backward()
            optimizer.step()

        model.eval()
        correct = 0
        total = 0
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
                preds = (torch.sigmoid(model(imgs).squeeze(1)) > 0.5).float()
                correct += (preds == labels).sum().item()
                total += labels.size(0)

        acc = correct / total
        print(f"Epoch {epoch+1}/{epochs} | Val Acc: {acc:.4f}")
        if acc > best_acc:
            best_acc = acc
            torch.save(model.state_dict(), "best_model.pt")

    model.load_state_dict(torch.load("best_model.pt", map_location=DEVICE))
    return model

def get_cnn_probs(model, val_tensors):
    model.eval()
    with torch.no_grad():
        return np.array([torch.sigmoid(model(t.unsqueeze(0).to(DEVICE))).item() for t in val_tensors])

def get_gradcam(model, img_tensor):
    saved = {}
    def fwd_hook(module, inp, out):
        saved["acts"] = out.detach()
    def bwd_hook(module, grad_in, grad_out):
        saved["grads"] = grad_out[0].detach()

    layer = model.layer4[-1]
    h1 = layer.register_forward_hook(fwd_hook)
    h2 = layer.register_full_backward_hook(bwd_hook)

    model.eval()
    out = model(img_tensor.unsqueeze(0).to(DEVICE))
    model.zero_grad()
    out.backward()

    acts = saved["acts"][0].cpu().numpy()
    grads = saved["grads"][0].cpu().numpy()

    weights = grads.mean(axis=(1, 2))
    cam = np.zeros(acts.shape[1:], dtype=np.float32)
    for i, w in enumerate(weights):
        cam += w * acts[i]
    cam = np.maximum(cam, 0)
    cam = cam / (cam.max() + 1e-8)
    cam = cv2.resize(cam, (224, 224))

    h1.remove()
    h2.remove()
    return cam

def show_gradcam(model, img_tensor, label, idx):
    cam = get_gradcam(model, img_tensor)
    orig = img_tensor.permute(1, 2, 0).numpy()
    orig = np.clip(orig * np.array([0.229, 0.224, 0.225]) + np.array([0.485, 0.456, 0.406]), 0, 1)

    heat = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
    heat = cv2.cvtColor(heat, cv2.COLOR_BGR2RGB) / 255.0
    overlay = 0.65 * orig + 0.35 * heat

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(orig)
    plt.title(f"Original ({'Bleached' if label == 1 else 'Healthy'})")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(cam, cmap="jet")
    plt.title("GradCAM")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(overlay)
    plt.title("Overlay")
    plt.axis("off")

    plt.tight_layout()
    plt.savefig(f"gradcam_{idx}.png", dpi=150)
    plt.show()

def evaluate(name, y_true, y_pred):
    print(f"\n{name}\n" + "=" * 40)
    print(classification_report(y_true, y_pred, target_names=["Healthy", "Bleached"]))
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Healthy", "Bleached"])
    disp.plot(cmap="Blues")
    plt.title(f"Confusion Matrix — {name}")
    plt.tight_layout()
    plt.savefig(f"cm_{name}.png", dpi=150)
    plt.show()

def get_spectral_features(img):
    img = img.astype(np.float32)
    R, G, B = img[..., 0], img[..., 1], img[..., 2]
    ExG = 2 * G - R - B
    rCBI = (R - B) / (R + B + 1e-5)
    NGRDI = (G - R) / (G + R + 1e-5)
    return [ExG.mean(), ExG.std(), rCBI.mean(), rCBI.std(), NGRDI.mean(), NGRDI.std()]

def train_xgb_from_csv(csv_path):
    df = pd.read_csv(csv_path)

    if "Bleaching_Severity_Int" in df.columns:
        y_tab = (pd.to_numeric(df["Bleaching_Severity_Int"], errors="coerce").fillna(0) > 0).astype(int)
    elif "Percent_Bleaching" in df.columns:
        y_tab = (pd.to_numeric(df["Percent_Bleaching"], errors="coerce").fillna(0) > 20).astype(int)
    else:
        raise ValueError("No suitable target column found")

    drop_cols = [c for c in ["Bleaching_Severity_Int", "Percent_Bleaching", "Date"] if c in df.columns]
    X_tab = df.drop(columns=drop_cols)

    for col in X_tab.columns:
        if X_tab[col].dtype == "object":
            X_tab[col] = X_tab[col].astype(str)

    X_tab = pd.get_dummies(X_tab, drop_first=True)
    X_tab = X_tab.apply(pd.to_numeric, errors="coerce")
    X_tab = X_tab.fillna(X_tab.mean(numeric_only=True))

    X_train_tab, X_val_tab, y_train_tab, y_val_tab = train_test_split(
        X_tab, y_tab, test_size=0.2, random_state=42, stratify=y_tab
    )

    xgb = XGBClassifier(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        eval_metric="logloss"
    )
    xgb.fit(X_train_tab, y_train_tab, eval_set=[(X_val_tab, y_val_tab)], verbose=False)

    preds = xgb.predict(X_val_tab)
    acc = (preds == y_val_tab).mean()
    print("XGBoost Accuracy:", acc)

    return xgb, X_tab

def fuse_predictions(p_cnn, p_xgb, alpha=0.6):
    return alpha * p_cnn + (1 - alpha) * p_xgb

def final_decision(p_cnn, p_xgb, threshold=0.5):
    fused = fuse_predictions(p_cnn, p_xgb)
    return int(fused > threshold), fused

def predict_cnn_prob(model, img_tensor):
    model.eval()
    with torch.no_grad():
        return torch.sigmoid(model(img_tensor.unsqueeze(0).to(DEVICE))).item()

def predict_xgb_prob(xgb_model, env_row, xgb_columns):
    row = pd.DataFrame([env_row])
    row = pd.get_dummies(row)
    row = row.reindex(columns=xgb_columns, fill_value=0)
    return xgb_model.predict_proba(row)[0, 1]

BASE = "../dataset"
CSV_PATH = "../dataset/coral_bleaching_cleaned.csv"

images, labels = load_images(BASE)

X_train_imgs, X_val_imgs, y_train, y_val = train_test_split(
    images, labels, test_size=0.2, random_state=42, stratify=labels
)

train_tensors = [train_transform(img) for img in X_train_imgs]
val_tensors = [val_transform(img) for img in X_val_imgs]

train_loader = torch.utils.data.DataLoader(
    list(zip(train_tensors, [torch.tensor(l, dtype=torch.float32) for l in y_train])),
    batch_size=16,
    shuffle=True
)

val_loader = torch.utils.data.DataLoader(
    list(zip(val_tensors, [torch.tensor(l, dtype=torch.float32) for l in y_val])),
    batch_size=16
)

model = build_model()
model = train_model(model, train_loader, val_loader)

for i in [0, 5, 10]:
    if i < len(val_tensors):
        show_gradcam(model, val_tensors[i], y_val[i], i)

cnn_probs = get_cnn_probs(model, val_tensors)
cnn_preds = (cnn_probs > 0.5).astype(int)

evaluate("CNN", np.array(y_val), cnn_preds)

xgb, xgb_columns = train_xgb_from_csv(CSV_PATH)

with open("xgb_model.pkl", "wb") as f:
    pickle.dump(xgb, f)

with open("xgb_columns.pkl", "wb") as f:
    pickle.dump(list(xgb_columns), f)

torch.save(model.state_dict(), "best_model.pt")

names = ["CNN", "XGBoost"]
accs = [None, None]

plt.figure(figsize=(6, 4))
plt.bar(names, [0.0, 0.0], color=["steelblue", "coral"])
plt.ylim(0, 1.0)
plt.ylabel("Accuracy")
plt.title("Model Comparison")
plt.tight_layout()
plt.savefig("model_comparison.png", dpi=150)
plt.show()
```
