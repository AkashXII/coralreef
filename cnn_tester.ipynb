```python
import os
import shutil
import random

SRC_DIR  = "d:\\CoralReef\\data\\clean_flat"  # change to your path
DST_DIR  = "d:\\CoralReef\\data\\clean"       # final train/val structure
VAL_SPLIT = 0.2
SEED      = 42

random.seed(SEED)

for cls in ["healthy", "bleached"]:
    src_cls = os.path.join(SRC_DIR, cls)

    images = [
        f for f in os.listdir(src_cls)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]
    random.shuffle(images)

    n_val   = int(len(images) * VAL_SPLIT)
    val_set = set(images[:n_val])

    for fname in images:
        split   = "val" if fname in val_set else "train"
        dst_cls = os.path.join(DST_DIR, split, cls)
        os.makedirs(dst_cls, exist_ok=True)
        shutil.copy(
            os.path.join(src_cls, fname),
            os.path.join(dst_cls, fname)
        )

    print(f"{cls}: {len(images) - n_val} train, {n_val} val")

print(f"\nDone. Final structure at: {DST_DIR}")
```
