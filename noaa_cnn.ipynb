```python
from huggingface_hub import hf_hub_download

model_path = hf_hub_download(
    repo_id="NMFS-OSI/yolo11n-cls-noaa-esd-coral-bleaching",
    filename="yolov11n-cls-noaa-esd-coral-bleaching-classifier.pt",
    local_dir="d:\\CoralReef\\models"
)

print(f"Model downloaded to: {model_path}")
```

```python
from ultralytics import YOLO

# Load the NOAA pretrained model
model = YOLO("d:\\CoralReef\\models1\\yolov11n-cls-noaa-esd-coral-bleaching-classifier.pt")

# Fine-tune on your coral dataset
results = model.train(
    data="d:\\CoralReef\\data\\clean",   # points at folder with train/ and val/
    task="classify",
    epochs=50,
    imgsz=224,                           # NOAA model was trained at 224x224
    batch=32,                            # RTX 3060 can handle this comfortably
    device=0,                            # GPU
    patience=10,                         # early stopping
    lr0=1e-4,                            # low learning rate — fine-tuning not training from scratch
    lrf=1e-6,                            # final learning rate
    warmup_epochs=3,
    augment=True,
    flipud=0.3,
    fliplr=0.5,
    degrees=15.0,
    hsv_h=0.015,
    hsv_s=0.3,
    hsv_v=0.3,
    dropout=0.3,                         # regularization — important with small dataset
    weight_decay=1e-4,
    project="d:\\CoralReef\\runs",
    name="noaa_finetuned",
    exist_ok=True,
    pretrained=True,
    verbose=True,
)

print("\nTraining complete")
print(f"Best weights: {results.save_dir}\\weights\\best.pt")
```
